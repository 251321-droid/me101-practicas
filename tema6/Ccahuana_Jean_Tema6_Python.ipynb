```python
import numpy as np
import matplotlib.pyplot as plt

# Degradado horizontal de 5x5
gris = np.tile(np.linspace(0, 255, 5), (5, 1))

print(gris)
print("shape:", gris.shape)
print("dtype:", gris.dtype)
print("mean :", gris.mean())
```

```python
plt.imshow(gris, cmap="gray", vmin=0, vmax=255)
plt.title("Degradado horizontal 5x5")
plt.colorbar()
plt.savefig("gris_tema6.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
recorte = gris[:, 2:]
print(recorte)
```

```python
rgb = np.zeros((4, 4, 3), dtype=np.uint8)
rgb[:, 0:2] = [255, 200, 0]   # mitad izquierda: amarillo-dorado
rgb[:, 2:4] = [0, 0, 255]     # mitad derecha: azul

print(rgb.shape)
print("Pixel [0,0]:", rgb[0, 0])
print("Pixel [0,3]:", rgb[0, 3])

plt.imshow(rgb)
plt.title("Imagen RGB de dos mitades")
plt.show()
```

```python
R = rgb[:, :, 0]
G = rgb[:, :, 1]
B = rgb[:, :, 2]

print("Canal R:\n", R)
print("Canal G:\n", G)
print("Canal B:\n", B)
```

```python
gris_promedio = rgb.mean(axis=2)
gris_luminosidad = 0.299 * R + 0.587 * G + 0.114 * B

print("Promedio simple:\n", gris_promedio)
print("Luminosidad ponderada:\n", gris_luminosidad)
```

```python
umbral = 100   # distinto de 128
binaria = (gris_luminosidad > umbral).astype(np.uint8)
print(binaria)

plt.imshow(binaria, cmap="gray")
plt.title(f"Binarización con umbral = {umbral}")
plt.show()
```
