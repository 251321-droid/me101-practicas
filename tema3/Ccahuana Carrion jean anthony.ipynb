```python
import numpy as np

notas = np.array([11, 14, 20, 8, 16, 13])

print("Shape:", notas.shape)
print("Dtype:", notas.dtype)
print("Ndim:", notas.ndim)
```

```python
promedio = notas.mean()
maximo = notas.max()
cantidad_aprobados = (notas >= 13.5).sum()

print("Promedio:", promedio)
print("Maximo:", maximo)
print("Aprobados:", cantidad_aprobados)
```

```python
import pandas as pd

datos = {
    "nombre": ["Ana", "Luis", "Carlos", "María", "Pedro"],
    "nota": [15, 9, 18, 12, 8],
    "asistencia_pct": [90, 80, 95, 75, 60]
}

df = pd.DataFrame(datos)

print(df)
```

```python
filtrado = df[(df["nota"] >= 13.5) & (df["asistencia_pct"] >= 70)]

print(filtrado)
```
