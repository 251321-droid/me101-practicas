```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("estudiantes_limpio.csv")
nota = df["nota"].dropna()

media = nota.mean()
mediana = nota.median()
std_muestral = nota.std()
var_poblacional = np.var(nota.to_numpy(), ddof=0)
q1 = nota.quantile(0.25)
q3 = nota.quantile(0.75)
iqr = q3 - q1

print(f"media={media:.4f}  mediana={mediana:.4f}")
print(f"std muestral (pandas)={std_muestral:.4f}  var poblacional (numpy)={var_poblacional:.4f}")
print(f"Q1={q1:.4f}  Q3={q3:.4f}  IQR={iqr:.4f}")
```

```python
skew_pd = nota.skew()
kurt_pd = nota.kurt()
skew_sp = stats.skew(nota)
kurt_sp = stats.kurtosis(nota)

print(f"pandas -> skew={skew_pd:.4f}  kurt={kurt_pd:.4f}")
print(f"scipy  -> skew={skew_sp:.4f}  kurt={kurt_sp:.4f}")
```

```python
def resumen_estadistico(serie, decimales=4):
    s = serie.dropna()
    desv = s.std()
    return {
        "n": int(s.shape[0]),
        "media": float(round(s.mean(), decimales)),
        "mediana": float(round(s.median(), decimales)),
        "desv_std": float(round(desv, decimales)),
        "cv_pct": float(round(desv / s.mean() * 100, decimales)),
    }

print(resumen_estadistico(df["nota"]))
```

```python
def clasificar_dispersion(cv_pct):
    if cv_pct < 15:
        return "Baja"
    elif cv_pct < 30:
        return "Moderada"
    else:
        return "Alta"

for columna in ["nota", "asistencia_pct"]:
    res = resumen_estadistico(df[columna])
    print(columna, res, "-> dispersión", clasificar_dispersion(res["cv_pct"]))
```
